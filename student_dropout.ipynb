## Question 1. Extract all students that study Online, have an attendance_rate smaller than 50, and have more than 5 missed_deadlines. Store them in a dataframe called at_risk.

```python

at_risk = df[
    (df["study_mode"] == "Online") &
    (df["attendance_rate"] < 50) &
    (df["missed_deadlines"] > 5)
]

at_risk
```

Out of the one hundred thousand students, 197 study online, have attendance below fifty, and have missed more than five deadlines.

## Question 2. Then compute, for each course_type, how many of those students there are and what fraction of them has dropped out. Store the result in a dataframe called at_risk_by_course, with columns students and dropout_rate, sorted by decreasing dropout_rate.

```python

# Count students in each course type
students = at_risk.groupby("course_type")["person_id"].count()

# Count students who dropped out
dropouts = at_risk.groupby("course_type")["dropout"].sum()

# Calculate dropout fraction rounded to 3 places
dropout_rate = (dropouts / students).round(3)

# Create the result DataFrame
at_risk_by_course = pd.DataFrame({
    "students": students,
    "dropout_rate": dropout_rate
})

# Sort by decreasing dropout rate
at_risk_by_course = at_risk_by_course.sort_values(
    "dropout_rate", ascending=False
)

at_risk_by_course
```

Among the 197 selected at-risk students, Law and Social Sciences has the highest dropout rate at 0.897, while Health Sciences has the lowest at 0.647. STEM and Business have the same dropout rate of 0.766. Law and Social Sciences has the highest proportion of dropouts, even though it doesn't have the highest number of students who dropped out.

## Question 3. Add to the dataframe the following columns: - pass_rate: the ratio between units_passed and units_enrolled; --gpa_change: the difference between gpa_semester_2 and gpa_semester_1;   --gpa_trend: a string that is "improving" if gpa_change is larger than 0.5, "declining" if gpa_change is smaller than -0.5, and "stable" otherwise.


```python

df["pass_rate"] = df["units_passed"] / df["units_enrolled"]# perform calculation inside dataframe with writing a loop- called vectorized calculation

df["gpa_change"] = df["gpa_semester_2"] - df["gpa_semester_1"]

df["gpa_trend"] = "stable" # first set everyone to stable 

df.loc[df["gpa_change"] > 0.5, "gpa_trend"] = "improving" # select rows where gpa change > 0.5 and update it

df.loc[df["gpa_change"] < -0.5, "gpa_trend"] = "declining"

df
```

I created three new columns. Pass rate equals units passed divided by units enrolled. GPA change equals semester two GPA minus semester one GPA. And GPA trend is improving if change is greater than 0.5, declining if less than negative 0.5, otherwise stable.

## Question 6. Partition the students into the following groups, according to weekly_work_hours: 0, 1-10, 11-20, 21-30, more than 30 (all intervals include both extremes).

```python
# Create the new column
df["work_hours_group"] = ""

# Students working zero hours
df.loc[df["weekly_work_hours"] == 0, "work_hours_group"] = "0"

# Students working 1 to 10 hours
df.loc[(df["weekly_work_hours"] >= 1) & (df["weekly_work_hours"] <= 10), "work_hours_group"] = "1-10"

# Students working 11 to 20 hours
df.loc[(df["weekly_work_hours"] >= 11) & (df["weekly_work_hours"] <= 20), "work_hours_group"] = "11-20"

# Students working 21 to 30 hours
df.loc[(df["weekly_work_hours"] >= 21) & (df["weekly_work_hours"] <= 30), "work_hours_group"] = "21-30"

# Students working more than 30 hours
df.loc[df["weekly_work_hours"] > 30, "work_hours_group"] = ">30"

df[["weekly_work_hours", "work_hours_group"]]

```

The output shows the original “weekly work hours” column and the new “work hours group” column we created. So, twenty-four hours falls into “21-30,” 16 falls into “11-20,” and so on

## Question 7.Build a risk score by summing:

## 2 points if the tuition is not up to date;1 point if stress_level is at least 8;1 point if sleep_hours is smaller than 5;1 point if the student has a part-time job with more than 20 weekly work hours;2 points if attendance_rate is smaller than 60;1 point if missed_deadlines is larger than 3. For each value of the risk score, compute the number of students and the dropout rate.

```python

df["risk_score"] = (
    2 * (df["tuition_up_to_date"] == 0)
    + (df["stress_level"] >= 8) # condition is true thats a +1, no need to add 1
    + (df["sleep_hours"] < 5)
    + (
        (df["part_time_job"] == 1)
        & (df["weekly_work_hours"] > 20)
    )
    + 2 * (df["attendance_rate"] < 60)
    + (df["missed_deadlines"] > 3)
)

# Count students for each risk score
students = df.groupby("risk_score")["person_id"].count()

# Calculate dropout rate for each risk score
dropout_rate = df.groupby("risk_score")["dropout"].mean().round(3)

# Combine the results into one table
risk_score_summary = pd.DataFrame({
    "students": students,
    "dropout_rate": dropout_rate
})

risk_score_summary
```

Output shows that as risk score increases, the dropout rate also increases. However, even at risk score zero, 15.3% of students still dropout, meaning the risk score does not account for every drop out.

```python
#verification of dropout rate
df[df["risk_score"] == 0]["dropout"].value_counts()
```

showing where values for mean of dropout rates were derived from. so 4120/26955 =0.153

## Question 8. Read the GDP per capita dataset published by Our World in Data directly from the URL

## https://ourworldindata.org/grapher/gdp-per-capita-worldbank.csv?v=1&csvType=full&useColumnShortNames=true and store it in a dataframe called gdp. The column ny_gdp_pcap_pp_kd contains the GDP per capita. Clean the dataset, removing the rows that do not correspond to a single country (e.g. World, High-income countries). For each country keep only the row with the most recent year.

```python
url='https://ourworldindata.org/grapher/gdp-per-capita-worldbank.csv?v=1&csvType=full&useColumnShortNames=true'

GDP = pd.read_csv(url)

GDP

```

GDP has a total of 7,445 rows and 5 columns

```python
GDP = GDP[(GDP["code"].str.len() == 3) | (GDP["code"] == "OWID_KOS")] 
GDP
```

To keep rows that correspond to a single country, i filtered rows with code of length 3 or codes owid_kos. Only 6,977 rows remains. Now we will extract only the recent available years per country

```python
GDP = GDP.sort_values("year", ascending=False)
GDP = GDP.drop_duplicates("code", keep="first")
GDP
```

After sorting the years in descending order and keeping the latest record for each country, 199 distinct countries remained in the GDP dataset.

## Question 9. Join the GDP data with the students data. Compute the dropout rate for each continent.

```python

students = df.copy() # make a copy of the original data
```

Making a copy for students

```python
# standardize USA to United states and UK to United Kingdom
students["nationality"] = students["nationality"].replace({
    "USA": "United States",
    "UK": "United Kingdom"
})
```

```python

students_GDP = students.merge(
    GDP,
    left_on="nationality",
    right_on="entity",
    how="left"
)

students_GDP
```

Left join GDP-entity on student-nationality

```python
# Count students for each continent
students_sg = students_GDP.groupby("owid_region")["person_id"].count()

# Calculate dropout rate for each continent
dropout_rate_sg = students_GDP.groupby("owid_region")["dropout"].mean().round(3)

# Combine the results into one table
continent_dropout_rate = pd.DataFrame({
    "students": students_sg,
    "dropout_rate": dropout_rate_sg
})

continent_dropout_rate = continent_dropout_rate.sort_values(
    "dropout_rate", ascending=False
)

continent_dropout_rate
```

There is little difference in dropout rates across the 6 continents, ranging from 31.7% in Africa to 32.3% in Asia and Europe. 

```python
#verification of dropout rate
students_GDP[students_GDP["owid_region"] == 'Africa']["dropout"].value_counts()
```

showing where values for mean of dropout rates per region was derived from. so 2504/7906 = 0.317
